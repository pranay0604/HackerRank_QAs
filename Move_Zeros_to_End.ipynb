## Move Zeros to End

You are given an array of integers. Your task is to move all the zeros to the end of the array, while maintaining the relative order of the non-zero elements. You must do this without creating a new array (in-place).

Input Format

First line: a single integer N — the number of elements in the array. 

Second line: N space-separated integers.

Constraints

1 ≤ N ≤ 10⁵ 

-10⁴ ≤ arr[i] ≤ 10⁴

Output Format

Print the modified array elements on a single line, separated by spaces.

Sample Input 0

5

0 1 0 3 12

Sample Output 0

1 3 12 0 0

Sample Input 1

1

7

Sample Output 1

7

Sample Input 2

5

1 2 3 0 0

Sample Output 2

1 2 3 0 0

Sample Input 3

6

0 1 0 2 0 3

Sample Output 3

1 2 3 0 0 0

```python
num = int(input().strip())
arr = list(map(int, input().split()))

pos = 0 

for i in range(num):
    if arr[i] != 0:
        arr[pos] = arr[i]
        pos += 1

for i in range(pos, num):
    arr[i] = 0

print(*arr)
```

```python
num = int(input().strip())
arr = list(map(int, input().split()))

pos = 0

for i in range(num):
    if arr[i] != 0:
        arr[pos], arr[i] = arr[i], arr[pos]
        pos += 1

print(*arr)
```
