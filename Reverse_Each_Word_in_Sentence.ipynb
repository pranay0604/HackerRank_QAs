## Reverse Each Word in a Sentence

Write a program that takes a sentence as input and reverses each word individually, while keeping the word order the same.

Instructions for the User

Take a sentence as input.

Reverse each word separately.

Print the modified sentence.

Input Format

A single line containing a sentence consisting of words separated by spaces.

Constraints

1≤length of sentence≤10 ^ 5 The sentence contains only alphabetic characters and spaces.

Output Format

Print the sentence where each word is reversed, but the order of the words remains the same.

Sample Input 0

hello world python

Sample Output 0

olleh dlrow nohtyp

Sample Input 1

I love coding

Sample Output 1

I evol gnidoc

```python
s = input().strip()

rev = " ".join(word[::-1] for word in s.split())

print(rev)
```

```python
sent = input()

words = sent.split()
result = []

for i in words:
    result.append(i[::-1])

print(" ".join(result))
```

```python
sentence = input()

words = sentence.split()
result = []

for word in words:
    rev = ""
    for char in word:
        rev = char + rev
    result.append(rev)

print(" ".join(result))
```

```python
sentence = input()

reversed_sentence = " ".join(["".join(reversed(word)) for word in sentence.split()])

print(reversed_sentence)
```
