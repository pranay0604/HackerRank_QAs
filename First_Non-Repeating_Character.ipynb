## First Non-Repeating Character

Write a program that takes a string as input and prints the first non-repeating character.

If every character repeats, print -1.

Input Format

A single line containing a string.

Constraints

The string contains lowercase English letters only.

Output Format

Print the first character that appears only once in the string. If no such character exists, print -1.

Sample Input 0

aabbcdde

Sample Output 0

c

Sample Input 1

aabbcc

Sample Output 1

-1

```python
s = input().strip()

freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

for ch in s:
    if freq[ch] == 1:
        print(ch)
        break
else:
    print(-1)
```

```python
from collections import Counter

s = input().strip()
count = Counter(s)

for ch in s:
    if count[ch] == 1:
        print(ch)
        break
else:
    print(-1)
```

```python
s = input().strip()

freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

result = [ch for ch in freq if freq[ch] == 1]

print(result[0] if result else -1)
```

```python
s = input().strip()

for ch in s:
    if s.count(ch) == 1:
        print(ch)
        break
else:
    print(-1)
```

```python
s = input().strip()

result = next((ch for ch in s if s.count(ch) == 1), -1)
print(result)
```
