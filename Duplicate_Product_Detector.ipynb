## Duplicate Product Detector

A warehouse inventory system stores product codes. Each code should appear exactly once in a list of N items, where every code is between 1 and N. 

Due to system errors, some codes appear twice. Your task is to detect all such duplicate codes. Approach: Use a visited-index trick — for each number val = abs(nums[i]), go to index val-1 and negate it. 

If that index is already negative, val is a duplicate. Alternatively, you may use a dictionary to count frequencies and collect those with count > 1. 

Output the duplicates in sorted ascending order, one per line. If no duplicates exist, print "No Duplicates". Example: Input: 8 4 3 2 7 8 2 3 1 Output: 2 3 (Both 2 and 3 appear twice)

Input Format

Line 1: Integer N 

Line 2: N space-separated integers (each in range [1, N])

Constraints

1 ≤ N ≤ 10^5 

1 ≤ nums[i] ≤ N

Output Format

Duplicate values in ascending order, one per line. If none, print "No Duplicates".

Sample Input 0

8

4 3 2 7 8 2 3 1

Sample Output 0

2

3

Sample Input 1

3

1 1 2

Sample Output 1

1

Sample Input 2

3

2 2 2

Sample Output 2

2

Sample Input 3

5

1 2 3 4 5

Sample Output 3

No Duplicates

```python
N = int(input().strip())
nums = list(map(int, input().split()))

freq = {}

for num in nums:
    freq[num] = freq.get(num, 0) + 1

duplicates = sorted([k for k, v in freq.items() if v > 1])

if duplicates:
    for d in duplicates:
        print(d)
else:
    print("No Duplicates")
```

```python
N = int(input().strip())
nums = list(map(int, input().split()))

duplicates = set()

for i in range(N):
    val = abs(nums[i])
    index = val - 1
    
    if nums[index] < 0:
        duplicates.add(val)
    else:
        nums[index] = -nums[index]

if duplicates:
    for d in sorted(duplicates):
        print(d)
else:
    print("No Duplicates")
```
