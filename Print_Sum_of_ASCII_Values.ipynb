## Print Sum of ASCII Values

Write a program to print sum of given string ascii values

Input Format

Take String input from the user

Constraints

Continue

Output Format

Print sum of ascii values

Sample Input 0

Python

Sample Output 0

642

Sample Input 1

ABCD

Sample Output 1

266

```python
text = input()

ascii_sum = 0

for ch in text:
    ascii_sum += ord(ch)

print(ascii_sum)
```

```python
text = input()

print(sum(ord(ch) for ch in text))
```

```python
text = input()

print(sum(map(ord, text)))
```

```python
text = input()

i = 0
total = 0

while i < len(text):
    total += ord(text[i])
    i += 1

print(total)
```

```python
def ascii_sum(s):
    if s == "":
        return 0
    return ord(s[0]) + ascii_sum(s[1:])

text = input()

print(ascii_sum(text))
```
