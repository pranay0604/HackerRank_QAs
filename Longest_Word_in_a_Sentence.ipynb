## Longest Word in a Sentence

Write a program that finds the longest word in a given sentence.

If multiple words have the same maximum length, return the first occurring word.

Input Format

A single line containing a sentence.

Constraints

Nothing

Output Format

Print the longest word in the sentence.

Sample Input 0

python is powerful language

Sample Output 0

powerful

Sample Input 1

I love programming

Sample Output 1

programming

```python
sentence = input()

words = sentence.split()

longest_word = words[0]

for word in words:
    if len(word) > len(longest_word):
        longest_word = word
        
print(longest_word)
```

```python
sentence = input()

longest = max(sentence.split(), key=len)

print(longest)
```

```python
sentence = input()
words = sentence.split()

max_length = max(len(word) for word in words)

longest = [word for word in words if len(word) == max_length]

print(longest[0])
```
