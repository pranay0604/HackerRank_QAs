## Float Arithmetic Formatter

Write a program that reads two floating-point numbers from the user and calculates their sum. The result must be printed rounded and formatted to exactly two decimal places.

Input Format

Read two floating-point numbers from the user in a single line, separated by a space.

Constraints

-10^6 ≤ float_value ≤ 10^6

Output Format

Print the sum of the two floating-point numbers formatted to exactly two decimal places.

Sample Input 0

3.456 2.123

Sample Output 0

5.58

Sample Input 1

-2.5 4.125

Sample Output 1

1.62

```python
a, b = map(float, input().split())

result = a + b

print(f"{result:.2f}")
```

```python
a, b = map(float, input().split())

result = round(a + b, 2)

print(f"{result:.2f}")
```

```python
a, b = map(float, input().split())

print("{:.2f}".format(a + b))
```
