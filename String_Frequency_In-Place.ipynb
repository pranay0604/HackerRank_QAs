## String Frequency In-Place

You are given a string s. For every character in s, count how many times that character appears consecutively starting from that position. Replace the character with that count. Return the resulting string of digits.

For example, in "abbc": — a at index 0 appears 1 time in a row → 1 — b at index 1 appears 2 times in a row → 2 — b at index 2 is still inside the same run of 2 → 2 — c at index 3 appears 1 time → 1 Result: "1221"

Each character is replaced by the total length of the consecutive run it belongs to — not its position within the run.

Input Format

Line 1: A single string s containing only lowercase English letters. No spaces. No extra lines. Read it using input().

Constraints

1 <= len(s) <= 10^5 s contains only lowercase English letters (a to z) 

No digit, space, or special character will appear in the input 

Each consecutive run can be at most 9 characters long (so output is always a single digit)

Output Format

Print a single string of digits on one line. 

The output string must have the same length as the input string. 

No spaces. No brackets. No commas.

Sample Input 0

abbc

Sample Output 0

1221

Sample Input 1

aaabbbccc

Sample Output 1

333333333

Sample Input 2

aabbcc

Sample Output 2

222222

Sample Input 3

aabbccdd

Sample Output 3

22222222

Sample Input 4

zzzzzzzzz

Sample Output 4

999999999

```python
s = input().strip()
n = len(s)

result = []
i = 0

while i < n:
    j = i
    
    while j < n and s[j] == s[i]:
        j += 1
    
    length = j - i
    
    result.append(str(length) * length)
    
    i = j  

print("".join(result))
```
