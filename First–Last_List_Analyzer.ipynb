## First–Last List Analyzer

Write a program that reads N integers from the user and stores them in a list. Your task is to print:

The first element of the list

The last element of the list

The length of the list

This problem checks your understanding of list indexing and the len() function.

Input Format

First line contains an integer N Second line contains N integers separated by space N a1 a2 a3 ... aN

Constraints

1 ≤ N ≤ 1000 

-10^6 ≤ ai ≤ 10^6

Output Format

first_element last_element length_of_list

Sample Input 0

4

10 20 30 40

Sample Output 0

10
40

4

Sample Input 1

2

100 200

Sample Output 1

100

200

2

```python
num = int(input())

numbers = list(map(int, input().split()))

print(numbers[0])       
print(numbers[-1])       
print(len(numbers))     
```

```python
N = int(input())
nums = input().split()

print(nums[0])
print(nums[N-1])
print(N)
```

```python
N = int(input())

nums = []
for x in input().split():
    nums.append(int(x))

print(nums[0])
print(nums[-1])
print(len(nums))
```

```python
N = int(input())
nums = list(map(int, input().split()))

first = nums[0]
last = nums[-1]
length = len(nums)

print(first)
print(last)
print(length)
```
