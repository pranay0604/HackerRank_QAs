## Move Zeros to the Front

You are given an integer n and a list of n integers.

Move every 0 in the list to the front. 

All non-zero numbers must appear after the zeros, in the same relative order they had in the input.

For example: Input: [1, 0, 2, 0, 3] 

Output: [0, 0, 1, 2, 3]

The zeros go to the front. The non-zero values 1, 2, 3 keep their original left-to-right order.

The relative order of non-zero elements must be preserved. Simply sorting the list will fail test cases that have negative numbers or repeated values.

Input Format

Line 1: A single integer n — the number of elements in the list. 

Line 2: n space-separated integers. Read both lines using input().

Constraints

1 <= n <= 10^5 

-10^4 <= each element <= 10^4 

The list may contain negative numbers 

The list may contain multiple zeros The list may have no zeros at all

Output Format

Print the resulting list as space-separated integers on a single line. 

No brackets. No commas. Just the numbers separated by spaces.

Sample Input 0

5

1 0 2 0 3

Sample Output 0

0 0 1 2 3

Sample Input 1

4

0 0 0 5

Sample Output 1

0 0 0 5

Sample Input 2

4

4 3 2 1

Sample Output 2

4 3 2 1

Sample Input 3

8

-1 0 3 0 -1 0 3 -1

Sample Output 3

0 0 0 -1 3 -1 3 -1

```python
n = int(input())
arr = list(map(int, input().split()))

zero = arr.count(0)

non_zero = [x for x in arr if x != 0]

result = [0] * zero + non_zero

print(*result)
```

```python
from collections import deque

n = int(input())
arr = list(map(int, input().split()))

dq = deque()

for x in arr:
    if x == 0:
        dq.appendleft(x)
    else:
        dq.append(x)

print(*dq)
```

```python
n = int(input())
arr = list(map(int, input().split()))

arr.sort(key=lambda x: x != 0)

print(*arr)
```

```python
n = int(input())
arr = list(map(int, input().split()))

end = n - 1  

for i in range(n - 1, -1, -1):
    if arr[i] != 0:
        arr[end] = arr[i]
        end -= 1

for i in range(end, -1, -1):
    arr[i] = 0

print(*arr)
```
