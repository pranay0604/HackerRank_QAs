## Exclusive Sum

You are given two sets A and B. Find the sum of all elements that exist in Set A but do not exist in Set B. If every element of A is also in B, the sum is 0.

Example: A = {1, 2, 3, 4} B = {3, 4, 5, 6} A - B = {1, 2} (elements only in A) Sum = 1 + 2 = 3 You must create actual Python sets. Use the set difference operation (A - B). Do not use loops to compare elements manually.

Input Format

Line 1: integer n1 — number of elements in Set A Line 2: n1 space-separated integers — elements of Set A Line 3: integer n2 — number of elements in Set B Line 4: n2 space-separated integers — elements of Set B

Constraints

1 <= n1, n2 <= 1000 

Elements are integers between -10^4 and 10^4 

If A is a subset of B — answer is 0 B may have elements not in A — ignore those

Output Format

Print a single integer — the sum of elements exclusive to Set A.

Sample Input 0

4

1 2 3 4

4

3 4 5 6

Sample Output 0

3

Sample Input 1

1 2 3

3

1 2 3

Sample Output 1

0

Sample Input 2

25 30

Sample Output 2

50

Sample Input 3

Sample Output 3

500
