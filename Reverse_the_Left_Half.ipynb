## Reverse the Left Half

You are given an array of N integers. Split the array into two halves. Reverse the left half (first N/2 elements) and keep the right half (remaining elements) unchanged. Print the final array. If N is odd, the middle element belongs to the right half and stays in place.

Input Format

First line: a single integer N — the number of elements. Second line: N space-separated integers.

Constraints

2 ≤ N ≤ 100000 

-10000 ≤ arr[i] ≤ 10000

Output Format

Print the final array on a single line with elements separated by spaces.

Sample Input 0

6

1 2 3 4 5 6

Sample Output 0

3 2 1 4 5 6

Sample Input 1

5

1 2 3 4 5

Sample Output 1

2 1 3 4 5

Sample Input 2

6

100 -100 0 50 -50 25

Sample Output 2

0 -100 100 50 -50 25

```python
num = int(input().strip())
arr = list(map(int, input().split()))

mid = num // 2

result = arr[:mid][::-1] + arr[mid:]

print(*result)
```

```python
N = int(input().strip())
arr = list(map(int, input().split()))

mid = N // 2

i, j = 0, mid - 1

while i < j:
    arr[i], arr[j] = arr[j], arr[i]
    i += 1
    j -= 1

print(*arr)
```

```python
N = int(input().strip())
arr = list(map(int, input().split()))

mid = N // 2

left_half = list(reversed(arr[:mid]))
right_half = arr[mid:]

print(*(left_half + right_half))
```

```python
N = int(input().strip())
arr = list(map(int, input().split()))

mid = N // 2

result = [arr[mid-1-i] for i in range(mid)] + arr[mid:]

print(*result)
```

```python
N = int(input().strip())
arr = list(map(int, input().split()))

mid = N // 2

result = []

for i in range(mid-1, -1, -1):
    result.append(arr[i])
    
for i in range(mid, N):
    result.append(arr[i])

print(*result)
```
