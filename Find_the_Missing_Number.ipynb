## Find the Missing Number

You are given an array of N-1 unique integers. The array contains numbers from 1 to N but exactly one number is missing. Your task is to find and print that missing number. You may not sort the array or use extra space — try to solve it in a single pass.

Input Format

First line: a single integer N — the upper bound of the range. 

Second line: N-1 space-separated integers.

Constraints

2 ≤ N ≤ 100000 1 ≤ arr[i] ≤ N 

All elements in the array are unique. 

Exactly one number from 1 to N is missing.

Output Format

Print a single integer — the missing number.

Sample Input 0

6

1 3 4 5 6

Sample Output 0

2

Sample Input 1

6

1 3 4 5 6

Sample Output 1

2

Sample Input 2

5

2 3 4 5

Sample Output 2

1

Sample Input 3

4

1 2 3 4

Sample Output 3

5

Sample Input 4

8

1 2 3 4 5 7 8

Sample Output 4

6

```python
N = int(input().strip())
arr = list(map(int, input().split()))

total_sum = N * (N + 1) // 2
actual_sum = sum(arr)

print(total_sum - actual_sum)
```

```python
N = int(input().strip())
arr = list(map(int, input().split()))

xor = 0

for i in range(1, N+1):
    xor ^= i

for num in arr:
    xor ^= num

print(xor)
```
