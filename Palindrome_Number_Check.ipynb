## Palindrome Number Check

A number is a palindrome if it reads the same forwards and backwards. For example, 121 is a palindrome but 123 is not. In this problem, you must check if a given integer N is a palindrome WITHOUT converting it to a string. Instead, reverse the number digit by digit using arithmetic operations and compare with the original. Algorithm:

Negative numbers are never palindromes. Store original value, then extract digits using % 10 and build reversed number using: rev = rev * 10 + digit, then N //= 10. If reversed number equals original, it is a palindrome.

Print "True" if palindrome, else "False". 

Example: Input: 121 Reversed: 121 == 121 → Output: True 

Input: -121 Negative → Output: False 

Input: 10 Reversed: 01 = 1 ≠ 10 → Output: False

Input Format

A single integer N (may be negative).

Constraints

-2^31 ≤ N ≤ 2^31 - 1

Output Format

Print "True" or "False".

Sample Input 0

121

Sample Output 0

True

Sample Input 1

-121

Sample Output 1

False

Sample Input 2

10

Sample Output 2

False

```python
N = int(input())


if N < 0:
    print("False")
else:
    original = N
    rev = 0

    while N > 0:
        digit = N % 10
        rev = rev * 10 + digit
        N = N // 10

    if rev == original:
        print("True")
    else:
        print("False")
```
