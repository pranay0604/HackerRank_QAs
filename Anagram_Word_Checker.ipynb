## Anagram Word Checker

A puzzle game gives players two words. The players win the round if the two words are anagrams of each other. Two words are anagrams if one word can be formed by rearranging the letters of the other — meaning both must have the exact same characters with the same frequency. Given two lowercase strings S1 and S2:

If they are anagrams, print YES Otherwise, print NO

Note: Case-insensitive comparison is NOT required here — both strings are already lowercase. 

Approach hint: Use a dictionary to count character frequencies in S1, then check if S2 has identical counts. If the lengths differ, immediately print NO. 

Example: 

Input: listen silent Output: YES (both have same letters: e,i,l,n,s,t each once) 

Input: 

hello 

world 

Output: NO

Input Format

Line 1: String S1 

Line 2: String S2

Constraints

1 ≤ len(S1), len(S2) ≤ 1000 

Strings contain only lowercase English letters.

Output Format

Print "YES" if anagrams, else "NO".

Sample Input 0

listen

silent

Sample Output 0

YES

Sample Input 1

hello

world

Sample Output 1

NO

Sample Input 2

aab

baa

Sample Output 2

YES

```python
from collections import Counter

s1 = input()
s2 = input()

if Counter(s1) == Counter(s2):
    print("YES")
else:
    print("NO")
```

```python
s1 = input()
s2 = input()

if len(s1) != len(s2):
    print("NO")
else:
    freq = {}

    for ch in s1:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in s2:
        if ch not in freq or freq[ch] == 0:
            print("NO")
            break
        freq[ch] -= 1
    else:
        print("YES")
```

```python
s1 = input()
s2 = input()

if len(s1) != len(s2):
    print("NO")
else:
    freq = [0] * 26

    for ch in s1:
        freq[ord(ch) - ord('a')] += 1

    for ch in s2:
        freq[ord(ch) - ord('a')] -= 1

    if all(x == 0 for x in freq):
        print("YES")
    else:
        print("NO")
```
