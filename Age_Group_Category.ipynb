## Age Group Category

Age < 13 → "Child"

13 ≤ Age ≤ 19 → "Teen"

20 ≤ Age ≤ 59 → "Adult"

Age ≥ 60 → "Senior Citizen"

Input Format

An integer age.

Constraints

0 ≤ age ≤ 120

Output Format

Category string.

Sample Input 0

16

Sample Output 0

Teen

Sample Input 1

65

Sample Output 1

Senior Citizen

```python
age = int(input().strip())

if age < 13:
    print("Child")
elif age <= 19:
    print("Teen")
elif age <= 59:
    print("Adult")
else:
    print("Senior Citizen")
```

```python
age = int(input().strip())

category = (
    "Child" if age < 13 else
    "Teen" if age <= 19 else
    "Adult" if age <= 59 else
    "Senior Citizen"
)

print(category)
```

```python
age = int(input().strip())

ranges = {
    "Child": range(0, 13),
    "Teen": range(13, 20),
    "Adult": range(20, 60),
    "Senior Citizen": range(60, 121)
}

for category, r in ranges.items():
    if age in r:
        print(category)
        break
```

```python
def get_category(age):
    if age < 13:
        return "Child"
    elif age <= 19:
        return "Teen"
    elif age <= 59:
        return "Adult"
    else:
        return "Senior Citizen"

age = int(input().strip())
print(get_category(age))
```

```python
age = int(input().strip())

conditions = [
    (age < 13, "Child"),
    (age <= 19, "Teen"),
    (age <= 59, "Adult"),
    (age >= 60, "Senior Citizen")
]

for cond, label in conditions:
    if cond:
        print(label)
        break
```
